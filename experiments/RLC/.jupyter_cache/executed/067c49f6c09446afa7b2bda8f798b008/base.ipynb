```python
import numpy as np
from scipy import signal
import matplotlib.pyplot as plt
```

```python
R = 100.0          # ohms
L = 1e-3           # henries
C = 10e-6          # farads
```

```python
num = [1.0]
den = [L*C, R*C, 1.0] # in descending order of degree

system = signal.TransferFunction(num, den)
```

```python
freq = 440
omega = freq*2*np.pi
t = np.linspace(0,4/freq,100)
V_in = signal.square(omega*t) 
tout, y, x = signal.lsim(system, V_in, t)
```

```python
plt.figure()
plt.plot(t, V_in, label="Input")
plt.legend()
plt.show()
```

```python
plt.figure()
plt.plot(t, y, label="Output")
plt.legend()
plt.show()
```
